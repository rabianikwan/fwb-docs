```python
import os
import sys
from pathlib import Path
import json
from pathlib import Path
from dotenv import load_dotenv
```

```python
# --------------- SETUP PATH & ENVIRONMENT ---------------
load_dotenv()
storage_folder = os.getenv("STORAGE_FOLDER")

current_directory = os.getcwd()

try:
	base_dir = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
except NameError:
	base_dir = os.path.dirname(os.path.dirname(current_directory))
sys.path.insert(0, base_dir)

path = Path.cwd()


parent = path.parent
storage = os.path.join(parent, storage_folder) # type: ignore

export_dir = os.path.join(storage, "skrining")
import_file = os.path.join(export_dir, "2.json")
output_file = os.path.join(export_dir, "skrining.csv")
```

```python
import csv

data = []
with open(import_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

print(data[0].keys())
```

```python
daftar_kunci = ['id', 'umur_tahun', 'sex', 'jumlah_kunjungan', 'lokasi_anatomi', 'diagnosa', 'protokol_wb', ]

with open(output_file, mode='w', newline='', encoding='utf-8') as berkas_csv:
    penulis_csv = csv.DictWriter(berkas_csv, fieldnames=daftar_kunci, extrasaction='ignore')
    
    penulis_csv.writeheader()
    
    for baris in data:
        penulis_csv.writerow(baris)
```
